```python
import pandas as pd

df = pd.read_csv("raw_breast_cancer_winsconsin_data.csv")
df.head(2)
```

```python
# Do some cleanup
df = df.drop(columns=['Unnamed: 32', 'id'], errors='ignore')
df.head(2)
```

```python
# Ecnode target feature
df['diagnosis'] = df['diagnosis'].map({'M': 1, 'B': 0})
df.head(2)
```

```python
# Dump the pre-processed data
df.to_csv('data.csv')
```
